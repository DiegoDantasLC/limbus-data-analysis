```python
import pandas as pd
import matplotlib.pyplot as plt


#Leitura
df = pd.read_csv(
    "limbus_ids_dados.csv",
    dtype={"rarity": str}
)
print("Primeira Linha")
print(df.info)


#Informações
print("\n Informações")
print(df.info())

print("\n Valores nulos")
print(df.isnull().sum())

print("\nQuantidade de IDs por Sinners")
print(df["sinner"].value_counts())


#Limpeza
df = df.dropna(subset=["rarity"])
df = df.drop_duplicates()
df["rarity"] = df["rarity"].replace("None", "000")


#Análise
ids_por_sinners = df["sinner"].value_counts()

print("\n IDs por Sinners: ")
print(ids_por_sinners)


#Gráfico
plt.figure(figsize=(14,5))
plt.bar(
    ids_por_sinners.index,
    ids_por_sinners.values
)

plt.title("Quantidade de IDs da Season 7 por Sinners")
plt.xlabel("Sinner")
plt.yticks([0,1,2])
plt.xticks(rotation=90)
plt.show()


#Exportação
df.to_csv("limbus_ids_limpar.csv", index=False)

print("\n O arquivo foi limpo e exportado com sucesso!")
```
